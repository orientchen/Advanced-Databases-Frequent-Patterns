# Text Retrieval: Term Weighting (TF, IDF, TF-IDF)

Worked examples corresponding to slides 16–19 of *Info_retrieval_01.pptx*. Run the cells in order. No additional Python packages are needed.

## 1. Term frequency weighting

Let $\mathrm{tf}(t,d)$ count the occurrences of term $t$ in document $d$. Three alternatives are:

**Binary:**
$$\mathrm{tfw}(t,d)=\begin{cases}1 & \mathrm{tf}>0\\0 & \mathrm{tf}=0\end{cases}$$

**Raw frequency:**
$$\mathrm{tfw}(t,d)=\mathrm{tf}(t,d)$$

**Normalized frequency:**
$$\mathrm{tfw}(t,d)=\begin{cases}c_1+c_2\frac{\mathrm{tf}(t,d)}{\mathrm{max\_tf}(d)} & \mathrm{tf}>0\\0 & \mathrm{tf}=0\end{cases}$$

Here $\mathrm{max\_tf}(d)$ is the maximum frequency of any term in $d$. The lecture uses $c_1=c_2=0.5$.

## 2. Numerical example (slide 17)

Document $d$ contains these terms and counts:

`huge: 1, information: 1, million: 1, people: 1, source: 1, web: 1, wide: 1, world: 2`

Thus $\mathrm{max\_tf}=2$ and:

$$\mathrm{tfw}(\text{huge},d)=0.5+0.5\frac{1}{2}=0.75$$
$$\mathrm{tfw}(\text{world},d)=0.5+0.5\frac{2}{2}=1.00$$

```python
term_counts = {'huge': 1, 'information': 1, 'million': 1, 'people': 1, 'source': 1, 'web': 1, 'wide': 1, 'world': 2}
max_tf = max(term_counts.values())
def normalized_tf(tf, max_tf, c1=0.5, c2=0.5):
    return c1 + c2 * tf / max_tf if tf > 0 else 0.0
for term, tf in term_counts.items():
    print(f'{term:12s} tf={tf}  normalized tfw={normalized_tf(tf, max_tf):.2f}')
assert normalized_tf(1, 2) == 0.75
assert normalized_tf(2, 2) == 1.0
```

## 3. Inverse document frequency (slide 18)

A term appearing in many documents receives less weight:

$$\mathrm{idfw}(t)=\log_{10}\left(\frac{N}{\mathrm{df}(t)}\right)$$

Here $N$ is the total number of documents and $\mathrm{df}(t)$ is the number containing term $t$. The slide uses $N=10$:

- `web`: $\mathrm{df}=5$, so $\log_{10}(10/5)\approx0.301$.
- `wide`: $\mathrm{df}=1$, so $\log_{10}(10/1)=1$.
- `world`: $\mathrm{df}=10$, so $\log_{10}(10/10)=0$.

The logarithm base is 10 here, matching the slide's numerical values.

```python
import math
N = 10
document_frequencies = {'web': 5, 'wide': 1, 'world': 10}
def idf(df, N):
    return math.log10(N / df)
for term, df in document_frequencies.items():
    print(f'{term:8s} df={df:2d}  idfw={idf(df, N):.3f}')
```

## 4. TF-IDF (slide 19)

Combine term frequency and inverse document frequency:

$$w(t,d)=\mathrm{tfw}(t,d)\times\mathrm{idfw}(t)$$

With normalized TF weighting:

$$w(t,d)=\left(0.5+0.5\frac{\mathrm{tf}(t,d)}{\mathrm{max\_tf}(d)}\right)\log_{10}\left(\frac{N}{\mathrm{df}(t)}\right)$$

For `web`, the slide specifies $\mathrm{tf}=1$, $\mathrm{max\_tf}=5$, $N=10$, and $\mathrm{df}=5$:

$$\mathrm{tfw}=0.5+0.5\frac{1}{5}=0.6$$
$$\mathrm{idfw}=\log_{10}(2)\approx0.301$$
$$w(\text{web},d)=0.6\times0.301\approx0.181\approx0.18$$

```python
tf, max_tf, df, N = 1, 5, 5, 10
tfw = normalized_tf(tf, max_tf)
idfw = idf(df, N)
weight = tfw * idfw
print(f'TF weight: {tfw:.3f}')
print(f'IDF weight: {idfw:.3f}')
print(f'TF-IDF: {weight:.3f} (approximately {weight:.2f})')
```
